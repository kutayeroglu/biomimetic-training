```python
import torch

from src.models import AlexNetModified
from src.eval.utils import calc_rf_indices
```

```python
# 1. Setup paths
model_path = (
    "/content/drive/MyDrive/colab_datasets/biomimetic_training/standard_checkpoint.pth"
)
```

```python
# 2. Load the model and its weights
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = AlexNetModified(num_classes=1000)
checkpoint = torch.load(model_path, map_location=device)
model.load_state_dict(checkpoint["model_state_dict"])
```

```python
# 3. Extract Conv1 weights (48 filters, 3 channels, 22x22)
conv1_weights = model.features[0].weight.data.cpu().numpy()
```

```python
# 4. Calculate the metrics (ensure return_rank=False to get raw values)
# The metrics are calculated as described in the paper's Supplemental Methods
color_metrics, freq_metrics, az_metrics = calc_rf_indices(
    conv1_weights, n_top_col_pixel=48, return_rank=False
)
```
