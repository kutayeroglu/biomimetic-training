```python
import os

import numpy as np
import torch

from src.models import AlexNetModified
from src.eval.utils import calc_rf_indices
from src.eval.viz.joint_color_freq import plot_fig_2_left

# 1. Setup paths and regimens
checkpoint_dir = "/content/drive/MyDrive/colab_datasets/biomimetic_training/"
cache_dir = os.path.join(checkpoint_dir, "metrics_cache")
os.makedirs(cache_dir, exist_ok=True)

regimens = {
    "Standard": "standard_checkpoint_epoch120.pth",
    "Biomimetic": "biomimetic_checkpoint_epoch120.pth",
    "Anti-Biomimetic": "anti_biomimetic_checkpoint_epoch120.pth",
}

# 2. Initialize the model once outside the loop
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = AlexNetModified(num_classes=1000) #
model.to(device)
model.eval()
```

```python
# 3. Iterate and load
for name, filename in regimens.items():
    cache_path = os.path.join(cache_dir, f"{name.lower()}_rf_indices.npz")
    full_model_path = os.path.join(checkpoint_dir, filename)

    if os.path.exists(cache_path):
        print(f"--- Loading cached metrics for {name} ---")
        data = np.load(cache_path)
        color_metrics = data['color']
        freq_metrics = data['freq']
        az_metrics = data['az']
    else:
        print(f"--- Loading weights for {name} from {full_model_path} ---")
        # Load the checkpoint and update the model state dict
        checkpoint = torch.load(full_model_path, map_location=device)
        model.load_state_dict(checkpoint["model_state_dict"]) #

        print(f"--- Calculating metrics for {name} ---")
        # Extract Conv1 weights
        # PyTorch format: [out_channels, in_channels, H, W]
        conv1_weights = model.features[0].weight.data.cpu().numpy()

        # Calculate metrics
        color_metrics, freq_metrics, az_metrics = calc_rf_indices(
            conv1_weights,
            n_top_col_pixel=48,
            return_rank=False
        )

        # Save results
        np.savez(cache_path, color=color_metrics, freq=freq_metrics, az=az_metrics)
        print(f"Saved metrics to {cache_path}")

    # 4. Proceed to plotting
    print(f"Generating plot for {name}...")
    plot_fig_2_left(color_metrics, freq_metrics, regimen_name=name)
```
